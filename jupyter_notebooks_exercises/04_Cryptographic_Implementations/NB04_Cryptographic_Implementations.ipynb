## Preparation

For shell command calls, the Python module ``subprocess`` is used (https://docs.python.org/3/library/subprocess.html).

Run the following code block to provide the functions ``execute_vm_command()`` and ``execute_stm32_command()`` within this notebook.

```python
import subprocess

def execute_vm_command(cmd):
    run = subprocess.run(cmd, universal_newlines=True, shell=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT)
    print(run.stdout)
    
def execute_stm32_command(cmd):
    ssh = 'sshpass -p "root" ssh root@192.168.13.13 '
    run = subprocess.run((ssh + cmd), universal_newlines=True, shell=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT)
    print(run.stdout)
```

## Exercise 4.1: OpenSSL AES Performance

### OpenSSL Speed Test

OpenSSL is one of the most popular cryptographic libraries on Linux systems. It also comes with the ``openssl`` command line tool that is helpful for many purposes.

The command ``openssl speed`` can be used to measure the performance of a variety of different algorithms on the given (embedded) platform. For specific tests, some special flags are useful:

`-evp`:
Use the specified cipher or message digest algorithm via the EVP interface.

`-elapsed`:
When calculating operations- or bytes-per-second, use wall-clock time instead of CPU user time as divisor. It can be useful when testing speed of hardware engines.

### Analysis of OpenSSL AES Software Implementation

In the first part of this exercise, the performance of the OpenSSL software implementation of the popular AES algorithm in CBC mode and with a 128-bit key has to be analyzed.

Execute the following command, wait for the results and reflect the results shown in the last lines of the output.

```python
execute_stm32_command('openssl speed -evp aes-128-cbc -elapsed')
```

## Exercise 4.2: OpenSSL SHA-256 Performance - Software vs. Hardware

### STM32MP157C Hardware Crypto Module

STM32MP157C SoCs include a cryptographic hardware module for block ciphers and hash functions. This allows to off-load crypto operations from the main CPU to this dedicated logic to save CPU time and energy. Also, it might be possible to improve performance.

ST provides a series of drivers for different cryptographic algorithms on the STM32MP1 platform.

``/proc/crypto`` contains a collection of available algorithms and operation modes. The STM32-specific ones start with ``stm32``, which can be listed by the following command.

```python
execute_stm32_command('cat /proc/crypto | grep stm32')
```

In order to enable the hardware crypto usage with OpenSSL, the kernel module ``cryptodev`` has to be loaded.

```python
execute_stm32_command('/sbin/modprobe cryptodev')
```

``lsmod`` lists the loaded kernel modules and, therefore, allows to check, which crypto modules are loaded correctly.

```python
execute_stm32_command('lsmod | grep cryp')
```

### Testing SHA-256 Performance

The availability of hardware support offers interesting advantages. However, it is hard to judge the performance of these co-processors without measuring them in a practical setting.

OpenSSL is able to use hardware engines, if the corresponding kernel modules are loaded.

Therefore, the following commands runs the same performance test for SHA-256, first in software, then in hardware.

```python
execute_stm32_command('openssl speed -evp sha-256 -elapsed')
```

```python
execute_stm32_command('openssl speed -elapsed -evp sha-256 -engine devcrypto')
```
